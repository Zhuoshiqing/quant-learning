## 3. 字符串与 f-string 格式化

- 字符串用单/双引号括起来，用 `+` 拼接；
- **f-string**：字符串前加 `f`，用 `{变量}` 插入内容；
- `:.2f` 保留两位小数，`:.2%` 按百分比显示两位小数。

```python
name = '贵州茅台'
price = 1500.25
ret = 0.0234   # 2.34%

print(name + ' 收盘价 ' + str(price))
print(f'{name} 收盘价 {price:.2f}')
print(f'今日涨跌幅 {ret:.2%}')
```

## 4. 条件判断 if / elif / else

注意冒号 `:` 和**缩进**（4 个空格）。程序按条件从上到下，满足一个就执行并跳出。

```python
ret = 0.05   # 当日涨跌幅

if ret >= 0.099:
    label = '涨停附近'
elif ret <= -0.099:
    label = '跌停附近'
elif ret > 0:
    label = '上涨'
elif ret < 0:
    label = '下跌'
else:
    label = '平盘'

print(label)
```

## 5. 循环：for、while 与 range / enumerate / zip

- `for x in 序列` 逐个遍历；`range(n)` 生成 0..n-1；
- `enumerate` 同时拿到序号和值；`zip` 把两个序列配对；
- `while 条件` 满足时反复执行。

```python
prices = [10.0, 10.2, 10.1, 10.3]

# range + 下标
for i in range(len(prices)):
    print(i, prices[i])

print('---')
# enumerate：序号和值一起拿
for i, p in enumerate(prices):
    print(f'第{i}天 价格{p}')

print('---')
# zip：配对
dates = ['周一', '周二', '周三']
for d_, p in zip(dates, prices):
    print(d_, p)
```

## 6. 列表与列表推导式

- 列表用 `[]`，可放任意类型；索引从 0 开始，`[-1]` 取最后一个；
- 切片 `a[1:3]`；`append` 末尾添加；
- **列表推导式**：`[对x的操作 for x in 序列 if 条件]`，用一行生成新列表。

```python
prices = [10.0, 10.2, 10.1, 10.3, 10.25]
print(prices[0], prices[-1])   # 第一个、最后一个
print(prices[1:4])             # 切片（不含第4个）

prices.append(10.4)            # 添加
print(prices)

# 由收盘价算每日收益率（pct_change 的手写版）
rets = [prices[i] / prices[i-1] - 1 for i in range(1, len(prices))]
print('收益率:', [round(r, 4) for r in rets])

# 筛选：跌幅超过 1% 的日子
down = [r for r in rets if r < -0.01]
print('跌幅>1%:', down)
```

**动手练 6**：给定收益率列表，

2. 再用推导式筛选出**上涨**（大于 0）的收益率。

```python
# 在这里写
rets = [0.02, -0.01, 0.015, 0.005, -0.02]



```

```python
# 参考答案
rets = [0.02, -0.01, 0.015, 0.005, -0.02]
pct = [r * 100 for r in rets]
up = [r for r in rets if r > 0]
print('百分点:', pct)
print('上涨的:', up)
```
